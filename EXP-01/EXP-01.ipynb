```python
import pandas as pd
data={"Name":["Alice","Bob","Charlie","David","Emma"],
      "Marks":["85","72","90","65","78"]}
df=pd.DataFrame(data)
print(df.iloc[1])
sort=df.sort_values(by="Marks", ascending=True)
print(sort)
d1=df["Marks"].tolist()
B=[]
for i in range(len(d1)):
    if int(d1[i])>=75:
        B.append("Pass")
    else:
        B.append("Fail")
df["Result"]=B
```

```python
import pandas as pd
data={"Name":["John","Sara","Mike","Emma"],
      "Salary":["25000","30000","28000","35000"]}
df=pd.DataFrame(data)
A=df["Salary"].tolist()
B=[int(i) for i in A]
for i in range(len(B)):
    B[i]=B[i]+B[i]*0.1
df["New_Salary"]=B
print(df)
```
